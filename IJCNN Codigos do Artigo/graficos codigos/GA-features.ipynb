```python
#8 FEATURES MLP

import matplotlib.pyplot as plt

#Dados
valores = [0.63234, 0.63487, 0.637732, 0.638936, 0.639056, 0.639432, 0.639899, 0.640832]
intervalos = [(0, 3), (4, 5), (6, 68), (69, 190), (191, 713), (714, 1168), (1169, 1845),(1846, 3000)]

x_points = []
y_points = []

for i, ((inicio, fim), valor) in enumerate(zip(intervalos, valores)):
    x_points.extend([inicio, fim])
    y_points.extend([valor, valor])
    if i < len(intervalos) - 1:
        proximo_valor = valores[i + 1]
        x_points.append(fim)
        y_points.append(proximo_valor)

plt.figure(figsize=(8,5))
plt.plot(x_points, y_points, color='#4da6ff', linewidth=2)

plt.title("Acurácia ao longo das gerações")
plt.xlabel("Gerações")
plt.ylabel("Acurácia")
plt.xlim(0, 1000)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
#18 FEATURES MLP

import matplotlib.pyplot as plt

# Dados atualizados
valores = [0.631471, 0.631743, 0.636287, 0.638575, 0.639297, 0.640019, 0.64026]
intervalos = [(0, 2), (3, 4), (5, 7), (8, 10), (11, 467), (486, 831), (832, 1000)]

# Criar listas para o gráfico em degraus
x_points = []
y_points = []

for i, ((inicio, fim), valor) in enumerate(zip(intervalos, valores)):
    # linha horizontal de inicio a fim
    x_points.extend([inicio, fim])
    y_points.extend([valor, valor])
    # se não for o último intervalo, adiciona a transição vertical
    if i < len(intervalos) - 1:
        proximo_valor = valores[i + 1]
        x_points.append(fim)
        y_points.append(proximo_valor)

# Plotar com azul claro
plt.figure(figsize=(8,5))
plt.plot(x_points, y_points, color='#4da6ff', linewidth=2)

# Configurações
plt.title("Acurácia ao longo das gerações")
plt.xlabel("Gerações")
plt.ylabel("Acurácia")
plt.xlim(0, 3000)
plt.grid(True)
plt.tight_layout()

plt.show()
```
